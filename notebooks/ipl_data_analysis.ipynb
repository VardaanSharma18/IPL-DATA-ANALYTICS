```python
# LOADING THE DATASETS
```

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/CSV_main")

matches = pd.read_csv(DATA_DIR / "matches.csv")
```

```python
matches_df = pd.DataFrame(matches)
```

```python
# Display first 5 rows:-
matches_df.head()
```

```python
# HANDLING MISSING VALUES
```

```python
print(matches_df.isnull().sum())
```

```python
numerical_cols = ["result_margin", "target_runs", "target_overs"]
categorical_cols_fill = ["city", "player_of_match", "winner", "method"]
 
for col in numerical_cols:
    matches_df[col].fillna(matches_df[col].mean(), inplace=True)      
 
for col in categorical_cols_fill:
    mode_val = matches_df[col].mode()
    if not mode_val.empty:
        matches_df[col].fillna(mode_val[0], inplace=True)    
```

```python
# REMOVING DUPLICATES
```

```python
matches_df.drop_duplicates(inplace=True)
```

```python
# DATA CLEANING AND TRANSFORMATION 
```

```python
# Converting categorical 'toss_decision' into numerical form 
le = LabelEncoder()
matches_df['toss_decision_num'] = le.fit_transform(matches_df['toss_decision'])
```

```python
# Renaming inconsistent columns 
matches_df.rename(columns={'player_of_match': 'Player_of_Match','id':'match_id'}, inplace=True)
```

```python
# Handling outliers in 'result_margin' using the IQR method 
Q1 = matches_df['result_margin'].quantile(0.25)
Q3 = matches_df['result_margin'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
matches_df = matches_df[(matches_df['result_margin'] >= lower_bound) & (matches_df['result_margin'] <= upper_bound)]
```

```python
# EXPLORATORY DATA ANALYSIS
```

```python
# Displaying column data types 
print("\nData types of columns:")
print(matches_df.dtypes)

# summary statistics 
print("\nSummary Statistics:")
print(matches_df.describe())

# Finding correlation between numerical features 
print("\nCorrelation Matrix:")
print(matches_df.corr(numeric_only=True))

```

```python
# DATA VISUALIZATION
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```
